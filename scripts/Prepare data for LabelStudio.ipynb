```python
import json
import pandas as pd
```

```python
samples_to_annotate = samples[samples['id'].isin(ids)]
samples_to_annotate = samples_to_annotate.sample(frac=1.0)
```

```python
samples_to_annotate.to_json("../data/texts_for_annotation.jsonl",orient='records',lines=True,force_ascii=False)
```
